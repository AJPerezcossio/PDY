ESCALA DE GRISES y PROFUNDIDAD DE COLOR

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import os

# 1. CARGA DE LA IMAGEN
# Cargamos en escala de grises (8 bits por defecto: uint8)
img = cv2.imread('windows.png', cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Error: No se encontró 'windows.png'. Verifica que el archivo esté en la misma carpeta.")
else:
    # 2. ANÁLISIS DE PROFUNDIDAD ORIGINAL
    # .dtype nos dice el tipo de dato (uint8 = 8 bits)
    print(f'Profundidad de color original: {img.dtype}')

    # 3. CONVERSIÓN A 16 BITS
    # Pasamos de un rango 0-255 a un rango 0-65535
    # Escalamos los valores para que la imagen no se vea negra al convertir
    img_16bits = np.uint16(img) * 257 
    print(f'Nueva profundidad de color: {img_16bits.dtype}')

    # 4. CREAR CARPETA DE SALIDA
    if not os.path.exists('img'):
        os.makedirs('img')

    # 5. GUARDAR RESULTADOS
    # OpenCV detecta automáticamente que es uint16 y guarda un PNG de 16 bits real
    cv2.imwrite('img/nueva16cv2.png', img_16bits)
    
    # Pyplot (Matplotlib) guarda lo que ve en el plot
    plt.imsave('img/nueva16plt.png', img_16bits, cmap='gray')

    # 6. VISUALIZACIÓN COMPARATIVA
    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.title(f'Original (8 bits - {img.dtype})')
    plt.imshow(img, cmap='gray')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.title(f'Convertida (16 bits - {img_16bits.dtype})')
    plt.imshow(img_16bits, cmap='gray')
    plt.axis('off')

    plt.show()
```

Crea una imagen en RGB que contenga 4 cuadrantes (CON COLORES RANDOM)

```python
import numpy as np
import matplotlib.pyplot as plt
import random as rd

tamanio_imagen = 100
mitad = tamanio_imagen // 2
num_aleatorio_a = rd.randint(0, 255)
num_aleatorio_b = rd.randint(0, 255)
num_aleatorio_c = rd.randint(0, 255)

# Crar 3 canales RGB de la imagen en blanco
R = np.ones((tamanio_imagen, tamanio_imagen)) * 255
G = np.ones((tamanio_imagen, tamanio_imagen)) * 255
B = np.ones((tamanio_imagen, tamanio_imagen)) * 255

# ciclo for para asignar colores
for i in range(tamanio_imagen):
    for j in range(tamanio_imagen):
        # AA
        if i < mitad and j < mitad:
            R[i, j] = num_aleatorio_a
            G[i, j] = 0
            B[i, j] = 0
        # AB
        elif i < mitad and j >= mitad:
            R[i, j] = 0
            G[i, j] = num_aleatorio_b
            B[i, j] = 0
        # BA
        elif i >= mitad and j < mitad:
            R[i, j] = 0
            G[i, j] = 0
            B[i, j] = num_aleatorio_c
        # BB
        else:
            R[i, j] = 0
            G[i, j] = 0
            B[i, j] = 0

imagen = np.dstack((R, G, B))
imagen = np.array(imagen, dtype=int)

plt.figure()
plt.axis('off')
plt.imshow(imagen)
plt.show()
```

RGB a HSV

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. CARGA DE LA IMAGEN
# Cargamos la imagen original de Windows
imagen1 = cv2.imread('windows.png', cv2.IMREAD_COLOR)

# Convertimos de BGR (OpenCV) a RGB (Estándar)
imagen1_rgb = cv2.cvtColor(imagen1, cv2.COLOR_BGR2RGB)

def RGB2HSV(img):
    """
    Convierte una imagen de espacio de color RGB a HSV.
    H (Hue): Matiz o tono (0-360 grados).
    S (Saturation): Saturación o pureza (0-1).
    V (Value): Valor o brillo (0-1).
    """
    # Convertimos a float32 y normalizamos de 0-255 a 0-1 para cálculos precisos
    img = img.astype(np.float32) / 255.0
    
    # Creamos una matriz vacía del mismo tamaño para guardar el resultado (H, S, V)
    new_img = np.zeros_like(img)

    # Extraemos información clave de cada píxel (R, G, B)
    maxChannel = np.argmax(img, axis=2) # Índice del canal con mayor valor (0=R, 1=G, 2=B)
    maxValue = np.amax(img, axis=2)    # Valor del canal máximo (esto define el Brillo/Value)
    minValue = np.amin(img, axis=2)    # Valor del canal mínimo
    diff = maxValue - minValue         # Diferencia entre el máximo y el mínimo (Croma)

    # --- CÁLCULO DEL CANAL V (VALUE / BRILLO) ---
    # El brillo es simplemente la intensidad del color más fuerte presente
    new_img[..., 2] = maxValue

    # --- CÁLCULO DEL CANAL S (SATURATION / SATURACIÓN) ---
    # Recorremos la imagen para calcular qué tan "puro" es el color
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if maxValue[i, j] == 0:
                new_img[i, j, 1] = 0 # Si el píxel es negro, la saturación es 0
            else:
                # La saturación es la relación entre la diferencia de canales y el valor máximo
                new_img[i, j, 1] = diff[i, j] / maxValue[i, j]

    # --- CÁLCULO DEL CANAL H (HUE / MATIZ) ---
    # El matiz define en qué parte del círculo cromático está el color
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            # Caso 1: Es un tono de gris (R=G=B), no tiene matiz
            if diff[i, j] == 0:
                new_img[i, j, 0] = 0
            
            # Caso 2: El Rojo es el canal predominante
            elif maxChannel[i, j] == 0: 
                new_img[i, j, 0] = 60 * ((img[i, j, 1] - img[i, j, 2]) / diff[i, j])
            
            # Caso 3: El Verde es el canal predominante
            elif maxChannel[i, j] == 1: 
                new_img[i, j, 0] = 120 + 60 * ((img[i, j, 2] - img[i, j, 0]) / diff[i, j])
            
            # Caso 4: El Azul es el canal predominante
            elif maxChannel[i, j] == 2: 
                new_img[i, j, 0] = 240 + 60 * ((img[i, j, 0] - img[i, j, 1]) / diff[i, j])

    # Corrección para ángulos negativos (aseguramos que el Hue esté entre 0 y 360°)
    mask_negativo = new_img[..., 0] < 0
    new_img[mask_negativo, 0] += 360
    
    return new_img

# 2. EJECUCIÓN DE LA FUNCIÓN
imagen_hsv = RGB2HSV(imagen1_rgb)

# 3. VISUALIZACIÓN DE RESULTADOS
fig, axs = plt.subplots(1, 4, figsize=(20, 6))

# Imagen Original
axs[0].imshow(imagen1_rgb)
axs[0].set_title('1. Original RGB')
axs[0].axis('off')

# Canal Hue (Tono) - Se visualiza mejor con el mapa de color 'hsv'
fig_h = axs[1].imshow(imagen_hsv[..., 0], cmap='hsv', vmin=0, vmax=360)
plt.colorbar(fig_h, ax=axs[1], shrink=0.4)
axs[1].set_title('2. Canal H (Hue)')

# Canal Saturation (Saturación)
fig_s = axs[2].imshow(imagen_hsv[..., 1], cmap='gray', vmin=0, vmax=1)
plt.colorbar(fig_s, ax=axs[2], shrink=0.4)
axs[2].set_title('3. Canal S (Saturation)')

# Canal Value (Brillo)
fig_v = axs[3].imshow(imagen_hsv[..., 2], cmap='gray', vmin=0, vmax=1)
plt.colorbar(fig_v, ax=axs[3], shrink=0.4)
axs[3].set_title('4. Canal V (Value)')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. CARGA DE IMAGEN
# Asegúrate de que la ruta sea correcta ('img/claseunopng.png' o 'windows.png')
imagen_bgr = cv2.imread('windows.png') 
imagen_rgb = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB)

# --- FUNCIÓN: RGB A HSV (Codificación) ---
def RGB2HSV(img):
    img = img.astype(np.float32) / 255.0
    new_img = np.zeros_like(img)

    maxChannel = np.argmax(img, axis=2)
    maxValue = np.amax(img, axis=2)
    minValue = np.amin(img, axis=2)
    diff = maxValue - minValue

    # Canal V (Value)
    new_img[..., 2] = maxValue

    # Canal S (Saturation)
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if maxValue[i, j] == 0:
                new_img[i, j, 1] = 0
            else:
                new_img[i, j, 1] = diff[i, j] / maxValue[i, j]

    # Canal H (Hue)
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if diff[i, j] == 0:
                new_img[i, j, 0] = 0
            elif maxChannel[i, j] == 0: # Rojo máximo
                new_img[i, j, 0] = 60 * ((img[i, j, 1] - img[i, j, 2]) / diff[i, j])
            elif maxChannel[i, j] == 1: # Verde máximo
                new_img[i, j, 0] = 120 + 60 * (img[i, j, 2] - img[i, j, 0]) / diff[i, j]
            elif maxChannel[i, j] == 2: # Azul máximo
                new_img[i, j, 0] = 240 + 60 * (img[i, j, 0] - img[i, j, 1]) / diff[i, j]

            if new_img[i, j, 0] < 0:
                new_img[i, j, 0] += 360
    return new_img

# --- FUNCIÓN: HSV A RGB (Decodificación) ---
def HSV2RGB(imagen_hsv):
    """
    Reconstruye la imagen RGB a partir de sus componentes HSV
    usando el modelo de geometría de color.
    """
    h_hsv = imagen_hsv.astype(np.float32)
    res_rgb = np.zeros_like(h_hsv)

    H = h_hsv[..., 0]
    S = h_hsv[..., 1]
    V = h_hsv[..., 2]

    # C = Croma (la intensidad del color puro)
    C = V * S
    # X = Valor intermedio basado en la posición en el hexágono de color
    X = C * (1.0 - np.abs(((H / 60.0) % 2.0) - 1.0))
    # m = Valor para ajustar el brillo (Value)
    m = V - C

    for i in range(h_hsv.shape[0]):
        for j in range(h_hsv.shape[1]):
            h = H[i, j]
            c = C[i, j]
            x = X[i, j]

            # El Hue (0-360) determina en qué sector del arcoíris estamos
            if 0 <= h < 60:
                rp, gp, bp = c, x, 0
            elif 60 <= h < 120:
                rp, gp, bp = x, c, 0
            elif 120 <= h < 180:
                rp, gp, bp = 0, c, x
            elif 180 <= h < 240:
                rp, gp, bp = 0, x, c
            elif 240 <= h < 300:
                rp, gp, bp = x, 0, c
            else: # 300 a 360
                rp, gp, bp = c, 0, x

            # Sumamos el ajuste 'm' para recuperar el brillo original
            res_rgb[i, j, 0] = rp + m[i, j]
            res_rgb[i, j, 1] = gp + m[i, j]
            res_rgb[i, j, 2] = bp + m[i, j]

    # Convertimos de 0-1 a 0-255 y a tipo entero de 8 bits
    return np.clip(res_rgb * 255.0, 0, 255).astype(np.uint8)

# 2. PROCESAMIENTO
img_hsv = RGB2HSV(imagen_rgb)
img_reconstruida = HSV2RGB(img_hsv)

# 3. MOSTRAR RESULTADOS
fig, axs = plt.subplots(1, 2, figsize=(15, 7))

# Nota: Mostrar HSV directo en Matplotlib se verá raro porque interpreta 
# los grados (0-360) como intensidades de color, por eso sale distorsionado.
axs[0].imshow(img_hsv) 
axs[0].set_title('Representación Matemática HSV')
axs[0].axis('off')

axs[1].imshow(img_reconstruida)
axs[1].set_title('Imagen Reconstruida en RGB')
axs[1].axis('off')

plt.show()
```

CONVERTIR IMAGEN DE HSV a RGB con las funciones

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. CARGA Y PREPARACIÓN
# Cargamos la imagen (BGR por defecto en OpenCV)
imagen_bgr = cv2.imread('windows.png') 
# Convertimos a RGB para que los colores sean naturales en Matplotlib
imagen_rgb = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB)

# --- FUNCIÓN: RGB A HSV (Codificación) ---
def RGB2HSV(img):
    img = img.astype(np.float32) / 255.0
    new_img = np.zeros_like(img)

    maxChannel = np.argmax(img, axis=2) # 0=R, 1=G, 2=B
    maxValue = np.amax(img, axis=2)
    minValue = np.amin(img, axis=2)
    diff = maxValue - minValue

    # Canal V (Value): Brillo máximo
    new_img[..., 2] = maxValue

    # Canal S (Saturation): Pureza del color
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if maxValue[i, j] == 0:
                new_img[i, j, 1] = 0
            else:
                new_img[i, j, 1] = diff[i, j] / maxValue[i, j]

    # Canal H (Hue): Tono en grados (0-360)
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if diff[i, j] == 0:
                new_img[i, j, 0] = 0
            elif maxChannel[i, j] == 0: # Rojo máximo
                new_img[i, j, 0] = 60 * ((img[i, j, 1] - img[i, j, 2]) / diff[i, j])
            elif maxChannel[i, j] == 1: # Verde máximo
                new_img[i, j, 0] = 120 + 60 * (img[i, j, 2] - img[i, j, 0]) / diff[i, j]
            elif maxChannel[i, j] == 2: # Azul máximo
                new_img[i, j, 0] = 240 + 60 * (img[i, j, 0] - img[i, j, 1]) / diff[i, j]

        # Corrección de ángulos negativos
        mask_neg = new_img[..., 0] < 0
        new_img[mask_neg, 0] += 360
        
    return new_img

# --- FUNCIÓN: HSV A RGB (Reconstrucción) ---
def HSV2RGB(imagen_hsv):
    imagen_hsv = imagen_hsv.astype(np.float32)
    res_rgb_norm = np.zeros_like(imagen_hsv)

    H, S, V = imagen_hsv[..., 0], imagen_hsv[..., 1], imagen_hsv[..., 2]

    # Cálculos matemáticos del modelo cilíndrico
    C = V * S
    X = C * (1.0 - np.abs(((H / 60.0) % 2.0) - 1.0))
    m = V - C

    for i in range(imagen_hsv.shape[0]):
        for j in range(imagen_hsv.shape[1]):
            h, c, x = H[i, j], C[i, j], X[i, j]

            if 0 <= h < 60:      r_p, g_p, b_p = c, x, 0
            elif 60 <= h < 120:  r_p, g_p, b_p = x, c, 0
            elif 120 <= h < 180: r_p, g_p, b_p = 0, c, x
            elif 180 <= h < 240: r_p, g_p, b_p = 0, x, c
            elif 240 <= h < 300: r_p, g_p, b_p = x, 0, c
            else:                r_p, g_p, b_p = c, 0, x

            res_rgb_norm[i, j, 0] = r_p + m[i, j]
            res_rgb_norm[i, j, 1] = g_p + m[i, j]
            res_rgb_norm[i, j, 2] = b_p + m[i, j]

    # Clip para asegurar rango 0-255 y conversión a uint8
    return np.clip(res_rgb_norm * 255.0, 0, 255).astype(np.uint8)

# 2. PROCESAMIENTO
imagen_hsv = RGB2HSV(imagen_rgb)
imagen_reconstruida = HSV2RGB(imagen_hsv)

# 3. SEPARACIÓN DE CANALES RGB
# Creamos imágenes que solo muestran un color (los otros canales en 0)
img_R = np.zeros_like(imagen_reconstruida)
img_R[:,:,0] = imagen_reconstruida[:,:,0]

img_G = np.zeros_like(imagen_reconstruida)
img_G[:,:,1] = imagen_reconstruida[:,:,1]

img_B = np.zeros_like(imagen_reconstruida)
img_B[:,:,2] = imagen_reconstruida[:,:,2]

# 4. VISUALIZACIÓN (Subplots 2x3)
fig, axs = plt.subplots(2, 3, figsize=(15, 10))

# Fila 1
axs[0,0].imshow(imagen_hsv) 
axs[0,0].set_title('Espacio Matemático HSV')
axs[0,1].imshow(imagen_reconstruida)
axs[0,1].set_title('Imagen RGB Reconstruida')
axs[0,2].axis('off') # Espacio vacío

# Fila 2: Canales individuales
axs[1,0].imshow(img_R)
axs[1,0].set_title('Canal Rojo (R)')
axs[1,1].imshow(img_G)
axs[1,1].set_title('Canal Verde (G)')
axs[1,2].imshow(img_B)
axs[1,2].set_title('Canal Azul (B)')

# Limpieza estética
for ax in axs.ravel():
    ax.axis('off')

plt.tight_layout()
plt.show()
```
