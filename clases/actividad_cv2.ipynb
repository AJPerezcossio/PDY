```python
import cv2
import matplotlib.pyplot as plt
import glob

# 1. Usamos glob para encontrar todos los archivos .jpeg en la carpeta actual
# Esto reemplaza el tener que escribir las rutas a mano una por una
archivos = glob.glob("*.jpeg")
archivos.sort() # Los ordena para que coincidan con Uno, Dos, Tres

# 2. Verificamos que existan al menos 3 imágenes para asignar a tus variables
if len(archivos) >= 3:
    # Cargamos las imágenes usando la lista que generó glob
    imagenUno = cv2.imread(archivos[0], cv2.IMREAD_COLOR)
    imagenDos = cv2.imread(archivos[1], cv2.IMREAD_COLOR)
    imagenTres = cv2.imread(archivos[2], cv2.IMREAD_COLOR)

    # Mensaje de confirmación (tu código original)
    print('Imagenes UNO, DOS y TRES cargadas correctamente usando glob')
    print(f"Archivo 1 ({archivos[0]}): {imagenUno.shape}")
    print(f"Archivo 2 ({archivos[1]}): {imagenDos.shape}")
    print(f"Archivo 3 ({archivos[2]}): {imagenTres.shape}")

    # 3. Visualización en RGB (Para que se vea bien en Visual Studio Code)
    plt.figure(figsize=(15, 5))

    # Imagen Uno
    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(imagenUno, cv2.COLOR_BGR2RGB))
    plt.title('Imagen UNO')
    plt.axis('off')

    # Imagen Dos
    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(imagenDos, cv2.COLOR_BGR2RGB))
    plt.title('Imagen DOS')
    plt.axis('off')

    # Imagen Tres
    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(imagenTres, cv2.COLOR_BGR2RGB))
    plt.title('Imagen TRES')
    plt.axis('off')

    plt.tight_layout()
    plt.show()

else:
    print(f"Error: Se necesitan 3 imágenes .jpeg, pero solo encontré {len(archivos)}")
```

```python
import cv2
import matplotlib.pyplot as plt
import glob
import os

# --- DEFINICIÓN DE LA FUNCIÓN (Tu lógica original) ---
def mostrar_imagen_completa(imagen, titulo):
    # Convertir a RGB para visualización correcta
    imagenRGB = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

    # Extraer canales individuales
    canalR = imagenRGB[:,:,0]
    canalG = imagenRGB[:,:,1]
    canalB = imagenRGB[:,:,2]

    # --- PRIMERA FILA: Descomposición de Canales ---
    plt.figure(figsize=(15,5))

    plt.subplot(1,5,1)
    plt.title(titulo + ' RGB')
    plt.imshow(imagenRGB)
    plt.axis('off')

    plt.subplot(1,5,2)
    plt.title('Canal R')
    plt.imshow(canalR, cmap='gray')
    plt.axis('off')

    plt.subplot(1,5,3)
    plt.title('Canal G')
    plt.imshow(canalG, cmap='gray')
    plt.axis('off')

    plt.subplot(1,5,4)
    plt.title('Canal B')
    plt.imshow(canalB, cmap='gray')
    plt.axis('off')

    plt.subplot(1,5,5)
    plt.title('Color Falso (Plasma)')
    plt.imshow(canalR, cmap='plasma')
    plt.axis('off')
    plt.show()

    # --- SEGUNDA FILA: Comparación con Barras de Color ---
    fig, axs = plt.subplots(1, 3, figsize=(15, 5))

    # Imagen Original
    axs[0].imshow(imagenRGB)
    axs[0].set_title(titulo + ' RGB')
    axs[0].axis('off')

    # Canal R con escala de intensidad
    img_gray = axs[1].imshow(canalR, cmap='gray')
    axs[1].set_title('Canal R (Escala Gris)')
    axs[1].axis('off')
    fig.colorbar(img_gray, ax=axs[1], shrink=0.8)

    # Color Falso Purples con escala
    img_falso = axs[2].imshow(canalR, cmap='Purples')
    axs[2].set_title('Color Falso (Purples)')
    axs[2].axis('off')
    fig.colorbar(img_falso, ax=axs[2], shrink=0.8)

    plt.show()

# --- CUERPO PRINCIPAL DEL PROGRAMA ---

# 1. Buscar archivos .jpeg automáticamente
archivos = glob.glob("*.jpeg")
archivos.sort()

# 2. Verificar y ejecutar
if len(archivos) < 3:
    print(f"Error: Se necesitan 3 imágenes, se encontraron {len(archivos)}")
else:
    # Cargar imágenes (usando los nombres reales encontrados)
    imagenUno = cv2.imread(archivos[0], cv2.IMREAD_COLOR)
    imagenDos = cv2.imread(archivos[1], cv2.IMREAD_COLOR)
    imagenTres = cv2.imread(archivos[2], cv2.IMREAD_COLOR)

    # Llamar a la función para cada imagen
    print(f"Procesando: {archivos[0]}")
    mostrar_imagen_completa(imagenUno, 'Imagen UNO')
    
    print(f"Procesando: {archivos[1]}")
    mostrar_imagen_completa(imagenDos, 'Imagen DOS')
    
    print(f"Procesando: {archivos[2]}")
    mostrar_imagen_completa(imagenTres, 'Imagen TRES')
```
