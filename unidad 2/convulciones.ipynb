```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_media_h(imagen, tkernel):
    kernel=np.ones(tkernel, np.float32)/np.prod(tkernel)
    imagen_c=cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img= cv2.imread(r'D:\UNIVALLE\PDY\unidad 1\clases\img\ImagenPrueba.jpeg', 0)  

img_fin= convolucion_media_h(img, 15)
plt.figure(figsize=(10,5))
plt.subplot(121)

plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.imshow(img_fin,cmap='gray')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((tkernel, tkernel), np.float32) / (tkernel * tkernel)
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img= cv2.imread(r'D:\UNIVALLE\PDY\unidad 1\clases\img\ImagenPrueba.jpeg', 1) 


img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


img_fin = convolucion_media_h(img_rgb, 10)


plt.figure(figsize=(10,5))

plt.subplot(121)
plt.title("Original")
plt.imshow(img_rgb)

plt.subplot(122)
plt.title("Filtrada")
plt.imshow(img_fin)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_media_vertical(imagen, alto_kernel):
    # Creamos un kernel de forma (alto, 1)
    # Por ejemplo: (15, 1). Esto lo hace estrictamente vertical.
    kernel = np.ones((alto_kernel, 1), np.float32) / alto_kernel
    
    # Aplicamos la convolución
    # El -1 indica que la imagen de salida tendrá la misma profundidad que la original
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

# Cargamos la imagen en escala de grises (0)
img = cv2.imread(r'D:\UNIVALLE\PDY\unidad 1\clases\img\ImagenPrueba.jpeg', 0)

if img is None:
    print("Error: No se encontró la imagen. Revisa la ruta.")
else:
    # Aplicamos el filtro con un tamaño de 30 para que el efecto vertical sea claro
    img_fin = convolucion_media_vertical(img, 30)

    # Configuración de la visualización
    plt.figure(figsize=(12, 6))

    # Imagen Original
    plt.subplot(121)
    plt.title("Original")
    plt.imshow(img, cmap='gray')
    plt.axis('off')

    # Imagen con Desenfoque Vertical
    plt.subplot(122)
    plt.title("Filtro Media Vertical")
    plt.imshow(img_fin, cmap='gray')
    plt.axis('off')

    plt.show()
```

diagonal 

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_diagonal(imagen, tkernel):
    # np.eye crea una matriz identidad (1s en la diagonal principal, 0s en el resto)
    # Dividimos solo por tkernel porque solo hay 'tkernel' cantidad de unos.
    kernel = np.eye(tkernel, dtype=np.float32) / tkernel
    
    # Aplicamos el filtro
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

# Cargar la imagen en escala de grises
img = cv2.imread(r'D:\UNIVALLE\PDY\unidad 1\clases\img\ImagenPrueba.jpeg', 0)

if img is not None:
    # Aplicamos el filtro diagonal con un tamaño de 20
    img_fin = convolucion_diagonal(img, 20)

    # Mostrar resultados
    plt.figure(figsize=(10, 5))
    
    plt.subplot(121)
    plt.title("Original")
    plt.imshow(img, cmap='gray')
    
    plt.subplot(122)
    plt.title("Efecto Diagonal ( \ )")
    plt.imshow(img_fin, cmap='gray')
    
    plt.show()
else:
    print("No se pudo cargar la imagen. Revisa la ruta.")
```

diagonal inversa

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_diagonal_inversa(imagen, tkernel):
    # np.eye(tkernel) crea la diagonal principal (\)
    # np.fliplr(...) la voltea horizontalmente para que sea inversa (/)
    kernel = np.fliplr(np.eye(tkernel, dtype=np.float32)) / tkernel
    
    # Aplicamos el filtro
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

# Cargar la imagen en escala de grises
img = cv2.imread(r'D:\UNIVALLE\PDY\unidad 1\clases\img\ImagenPrueba.jpeg', 0)

if img is not None:
    # Aplicamos el filtro con un tamaño de 20 para notar el barrido
    img_fin = convolucion_diagonal_inversa(img, 20)

    # Visualización de resultados
    plt.figure(figsize=(10, 5))
    
    plt.subplot(121)
    plt.title("Original")
    plt.imshow(img, cmap='gray')
    
    plt.subplot(122)
    plt.title("Diagonal Inversa ( / )")
    plt.imshow(img_fin, cmap='gray')
    
    plt.show()
else:
    print("Error: No se pudo cargar la imagen.")
```

en una funcion 

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def aplicar_filtro_media(imagen, tam, tipo):
    """
    tipo 1: Horizontal
    tipo 2: Vertical
    tipo 3: Diagonal (\)
    tipo 4: Diagonal Inversa (/)
    """
    if tipo == 1: # Horizontal
        kernel = np.ones((1, tam), np.float32) / tam
    elif tipo == 2: # Vertical
        kernel = np.ones((tam, 1), np.float32) / tam
    elif tipo == 3: # Diagonal
        kernel = np.eye(tam, dtype=np.float32) / tam
    elif tipo == 4: # Diagonal Inversa
        kernel = np.fliplr(np.eye(tam, dtype=np.float32)) / tam
    else:
        print("Tipo no válido")
        return imagen

    # Aplicar la convolución
    return cv2.filter2D(imagen, -1, kernel)

# --- PRUEBA DEL CÓDIGO ---
img = cv2.imread(r'D:\UNIVALLE\PDY\unidad 1\clases\img\ImagenPrueba.jpeg', 0)

if img is not None:
    # AQUÍ ELIGES EL NÚMERO (1, 2, 3 o 4)
    opcion = 3 
    resultado = aplicar_filtro_media(img, 25, opcion)

    plt.figure(figsize=(10,5))
    plt.subplot(121), plt.imshow(img, cmap='gray'), plt.title("Original")
    plt.subplot(122), plt.imshow(resultado, cmap='gray'), plt.title(f"Filtro Tipo {opcion}")
    plt.show()
```
